```python
import pandas as pd

df = pd.read_excel("startpop_norge.xlsx")
df
```

```python
# Funksjon for å ekspandere DataFrame
def expand_rows(row):
    menn = pd.DataFrame({'Alder': [row['Alder']] * row['Menn'], 'Kjønn': ['Menn'] * row['Menn']})
    kvinner = pd.DataFrame({'Alder': [row['Alder']] * row['Kvinner'], 'Kjønn': ['Kvinner'] * row['Kvinner']})
    return pd.concat([menn, kvinner])

# Ekspandere DataFrame
expanded_df = pd.concat([expand_rows(row) for _, row in df.iterrows()], ignore_index=True)

```

```python
expanded_df["alder"] = expanded_df["Alder"].str.split(" ").str[0].astype(int)
expanded_df
```
