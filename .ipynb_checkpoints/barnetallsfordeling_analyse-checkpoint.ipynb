```python
import pandas as pd
from pathlib import Path
import os
```

```python
barnefordeling_path = Path("data_barnefordeling")
```

```python
files = sorted([file for file in os.listdir(barnefordeling_path) if file.endswith("Kvinnen.csv")])
files
```

```python
files = sorted([file for file in os.listdir(barnefordeling_path) if file.endswith(".csv")])
files
```

```python
for file in files:
    df = pd.read_csv(barnefordeling_path / file)
```

```python
df_full = pd.DataFrame({"årstall" :list(range(1935, 2004, 5)) + [2004] })
col = "4 barn eller flere"
for file in files:
    df = pd.read_csv(barnefordeling_path / file)
    df_fire_barn = df[["årstall",col]]
    df_fire_barn.loc[:,col] = df_fire_barn[col].str.replace(",",".").astype(float)
    df_fire_barn.columns = ["årstall", df["kjønn"].unique()[0] + " " + str(df["alder"].unique()[0])]
    df_full = df_full.merge(df_fire_barn, on ="årstall", how="left") 
df_full.plot(x="årstall")
```
