```python
import pandas as pd
import os
import matplotlib.pyplot as plt

[file for file in os.listdir(os.getcwd()) if file.endswith(".xlsx") or file.endswith(".csv")]
```

```python
fodsler = pd.read_excel("fødte.xlsx")
columns = ["år"]
columns.extend(fodsler.columns[1:])
fodsler.columns = columns#.str.strip().str.replace(" ", "_")
fodsler.columns = fodsler.columns.str.strip().str.lower().str.replace(" ", "_")
fodsler
```

```python
befolkning = pd.read_excel('befolkning_m_k.xlsx').rename(columns = {"Alder":"alder"})
befolkning.kjonn = befolkning.kjonn.ffill()
befolkning
```

# Men and women

```python
df = befolkning
# Separating the data for plotting
menn = df[df['kjonn'] == 'Menn']
kvinner = df[df['kjonn'] == 'Kvinner']

# Plotting
fig, ax = plt.subplots(figsize=(10, 8))

ax.barh(menn['alder'], menn['2023'], color='blue', alpha=0.7, label='Men')
ax.barh(kvinner['alder'], -kvinner['2023'], color='orangered', alpha=0.7, label='Women')

ax.set_xlabel('Population')
ax.set_title('Population Distribution by Age Group and Gender (2023)')
ax.legend()

# Customizing the x-axis to show positive values for both sides
ticks = ax.get_xticks()
ax.set_xticklabels([int(abs(tick)) for tick in ticks])

plt.savefig("pop_distribution_2024.png", format="png", dpi=300)

plt.show()
```

```python
befolkning = befolkning.loc[lambda x: x.kjonn == "Kvinner"]
befolkning
```

```python
fruktbare = befolkning.iloc[3:10,:]#.sum(axis=0).reset_index().iloc[1:].rename(columns={"index":"år",0:"bef"})
fruktbare
```

```python
# fodte
fodsler = pd.read_excel('levendefodte_etter_alder.xlsx')
fodsler["alder"] = fodsler["Alder"].str.split(" ").str[0].astype(int)
fodsler.drop("Alder", axis=1, inplace=True)
fodsler
```

```python
bins = [15, 20, 25, 30, 35, 40, 45, 50]
labels = ['15-19 år', '20-24 år', '25-29 år', '30-34 år', '35-39 år', '40-44 år', '45-49 år']

# Bin the data
fodsler['aldersgruppe'] = pd.cut(fodsler['alder'], bins=bins, labels=labels, right=False)
fodsler
```

```python
fodsler_gruppe = fodsler.groupby("aldersgruppe").sum().drop("alder", axis=1).reset_index()
fodsler_gruppe
```

```python
aar = "2023"
```

```python
# Vi sjekker ut TFR for 2023
asfr = pd.DataFrame()
asfr["aldersgruppe"] = fodsler_gruppe.aldersgruppe

for i in range(2017, 2024):
    aar = str(i)
    fruktbare_2023 = fruktbare.loc[:, ["alder", aar]].rename(columns={aar:"befolkning"})
    fodsler_gruppe_2023 = fodsler_gruppe.loc[:, ["aldersgruppe", aar]].rename(columns={"aldersgruppe":"alder", aar:"fodte"})
    fruktbare_2023

    fruktbare_og_fodsler = fruktbare_2023.merge(fodsler_gruppe_2023, on ="alder", how="left")

    asfr[f"asfr_{aar}"] = fruktbare_og_fodsler["fodte"] / (fruktbare_og_fodsler["befolkning"] / 1000)

    # asfr[f"asfr_{aar}"] = (asfr[f"asfr_{aar}"] / asfr[f"asfr_{aar}"].sum())
    
asfr.set_index("aldersgruppe", inplace=True)
asfr
```

```python
asfr.sum()
```

```python
(asfr.sum(axis=1) / len(asfr.columns)).reset_index().rename(columns={0:"asfr_avg"})
asfr.to_csv("asfr_avg.csv")
```

```python
import matplotlib.pyplot as plt
import pandas as pd

# Sample data for illustration
df = fruktbare_og_fodsler

# Plotting
fig, ax1 = plt.subplots(figsize=(10, 8))

# Plotting population
ax1.bar(df['alder'], df['befolkning'], color='orangered', alpha=0.3, label='Population')
ax1.set_xlabel('Age Group')
ax1.set_ylabel('Num. of people', color='orangered')
ax1.tick_params(axis='y', labelcolor='orangered')
ax1.set_xticks(range(len(df['alder'])))
ax1.set_xticklabels(df['alder'])

# Creating a second y-axis for live births
ax2 = ax1.twinx()
ax2.scatter(df['alder'], df['fodte'], color='blue', label='Live births')
ax2.set_ylabel('Num. of live births', color='blue')
ax2.tick_params(axis='y', labelcolor='blue')

# Adding title and legend
fig.suptitle('Population and live births for different age groups in Norway (2024)')
fig.tight_layout()

# Save the plot as an image file
# plt.savefig("ASFR_2024.png", format="png", dpi=300)

plt.show()
```

```python
import numpy as np 

df = fruktbare_og_fodsler

# Define the positions of the bars
x = np.arange(len(df['alder']))
width = 0.4  # Width of the bars

# Plotting
fig, ax1 = plt.subplots(figsize=(10, 8))

# Plotting population bars
ax1.bar(x - width/1.9, df['befolkning'], width, color='orangered', alpha=0.7, label='Population')

# Plotting live births bars
ax1.bar(x + width/1.9, df['fodte'], width, color='blue', alpha=0.7, label='Live births')

# Customizing the x-axis
ax1.set_xlabel('Age Group')
ax1.set_ylabel('Number of People')
ax1.set_title('Population and Live Births for Different Age Groups in Norway (2023)')
ax1.set_xticks(x)
ax1.set_xticklabels(df['alder'])

# Adding legend
ax1.legend()

# Adjust layout
fig.tight_layout()

# Save the plot as an image file
plt.savefig("pop_and_births.png", format="png", dpi=300)

# Display the plot
plt.show()
```
