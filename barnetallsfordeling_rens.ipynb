```python
def fix_barnetallsfordeling():
    barnetallsfordeling = pd.read_csv(data_path / "barnetallsfordeling.csv",skiprows=1, nrows=16, sep=";").reset_index(drop=True)
    info = barnetallsfordeling.iloc[0,0]
    kjønn = info.split(" ")[0]
    alder = int(info.split(" ")[-2])
    alder
    barnetallsfordeling["kjønn"] = kjønn.replace("s", "")
    barnetallsfordeling["alder"] = alder
    barnetallsfordeling.columns = ["årstall", "0 barn", "1 barn", "2 barn", "3 barn", "4 barn eller flere", "gjennomsnittlig barnetall", "kjønn", "alder"]
    barnetallsfordeling = barnetallsfordeling.iloc[1:]
    return barnetallsfordeling
```
